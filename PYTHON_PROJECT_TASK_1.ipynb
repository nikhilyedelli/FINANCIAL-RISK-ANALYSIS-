**Task 1: Data Cleaning and Formatting (10 Marks)**

**1.1** Remove/treat any special characters or non-numeric entries from financial fields.

```python
#We need to remove everything except digits and decimal points.
financial_cols = ["TransactionAmount", "AccountBalance", "RiskScore", "CreditRating", "TenureMonths"]

for col in financial_cols:
    df[col] = df[col].astype(str)

    for i in range(len(df)):
        df.at[i, col] = re.sub(r'[^0-9.]', '', df.at[i, col])

    df[col] = pd.to_numeric(df[col], errors='coerce')
```

**1.3** Validate and Format Date Columns

```python
import pandas as pd

df = pd.read_csv("/content/morgan_stanely.csv")
```

```python
# Validate the date column
# Your dates are in DD-MM-YYYY format, so use:
df["TransactionDate"] = pd.to_datetime(
    df["TransactionDate"],
    format="%d-%m-%Y"
)
```

```python
#Check if any invalid dates exist
print(df["TransactionDate"].isnull().sum())
```

```python
#Format the dates
df["TransactionDate"] = df["TransactionDate"].dt.strftime("%Y-%m-%d")
```
